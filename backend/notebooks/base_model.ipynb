```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

X = df[feature_cols].fillna(0).astype('float32').values  # float32 halves memory too

def get_similar(player_idx, X, top_k=10):
    query = X[player_idx].reshape(1, -1)
    sims = cosine_similarity(query, X)[0]
    top_indices = np.argsort(sims)[::-1]
    top_indices = top_indices[top_indices != player_idx][:top_k]  # exclude itself
    return df.iloc[top_indices][['short_name', 'overall']], sims[top_indices]
```
